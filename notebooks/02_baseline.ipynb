```python
import pandas as pd, matplotlib.pyplot as plt
from PIL import Image
idx = pd.read_csv("results/generations/index.csv")
b = idx[(idx.config == "base") & (idx.seed == 0)]
fig, axes = plt.subplots(2, (len(b) + 1) // 2, figsize=(20, 7))
for ax, (_, r) in zip(axes.flat, b.iterrows()):
    ax.imshow(Image.open(r.file)); ax.set_title(r.prompt[:40], fontsize=7); ax.axis("off")
plt.tight_layout(); plt.savefig("results/figures/baseline.png", dpi=120)
```
